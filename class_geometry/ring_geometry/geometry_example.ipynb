# Example use of the Geometry class

```python
"""Example use of the Geometry class"""

import numpy as np # type: ignore
from rcsparameters.ring_geometry.ring_geometry import RingGeometry
```

Create a Geometry object and print the input parameters chosen in the file_input

```python
file_input = 'data_geometry/para_RCS_ME.json'
RCS_geom = RingGeometry(file_input)

print('Input parameters')
print('Circumference [m]', RCS_geom.circumference)
print('Injection energy [GeV]', RCS_geom.injection_energy / 1e9)
print('Extraction energy [GeV]', RCS_geom.extraction_energy / 1e9)
print('Length of the straight insertion section [m]', RCS_geom.length_short_insertion)
print('Dipole spacing [m]', RCS_geom.dipole_spacing)
print('Number of arcs', RCS_geom.nb_arc)
print('Number of insertion', RCS_geom.nb_RF_section)
print('Number of cells per arc', RCS_geom.nb_cell_arc)
print('Dipole properties', RCS_geom.dipoles)
print('Dipole pattern of a half-cell', RCS_geom.pattern)
print('Filling ratio of the arcs', RCS_geom.filling_ratio)
print('Filling ratio of the RF', RCS_geom.filling_ratio_RF)
```

See some geometric parameters computed by the class

```python
print('Total arc length [m]', RCS_geom.tot_arc_length)
print('Arc length [m]', RCS_geom.arc_length)
print('Total insertion length [m]', RCS_geom.tot_insertion_length)
print('Insertion length [m]', RCS_geom.insertion_length)
print('Cell length [m]', RCS_geom.cell_length)
print('Total length of dipole', RCS_geom.dipole_length_tot)
print('Dipole filling factor', RCS_geom.filling_ratio_dipole)
print('Dipole parameters', RCS_geom.dipole_families)
print('Total trajectory excursion [mm]', (np.max(RCS_geom.extrema(RCS_geom.t_ext)[1]
            ) - np.min(RCS_geom.extrema(RCS_geom.t_inj)[0])) * 1e3)
mask_nc = [item == 'BNC' for item in RCS_geom.pattern]
mask_sc = [item == 'BSC' for item in RCS_geom.pattern]
max_width_dip = RCS_geom.extrema(RCS_geom.t_ext)[1] - RCS_geom.extrema(RCS_geom.t_inj)[0]
print('Max width in SC [mm]', np.max(max_width_dip[mask_sc]) * 1e3)
print('Max width in NC [mm]', np.max(max_width_dip[mask_nc]) * 1e3)
```

Some methods provide parameters depending on time. Note that the time is normalised between 0 and 1. 

```python
t_print = 0.5
y_min, y_max, width, max_apert_noshift, max_apert = RCS_geom.extrema(t_print)
print('Dipole angle', RCS_geom.theta(t_print))
print('Width in dipoles [mm]', width * 1e3)
print('Max apert no shift [mm]', max_apert_noshift * 1e3)
print('Path length [m]', RCS_geom.path_length(t_print))
print('Path length tot [m]', RCS_geom.path_length_tot(t_print))
```

The method plot_traj plots the trajectory in a cell for different times. The method plot_path_length_diff plots the path length variation. 

```python
n_traj = 8
t_traj = np.linspace(RCS_geom.t_inj, RCS_geom.t_ext, n_traj)
fig_traj1 = RCS_geom.plot_traj(t_traj)
fig_path_length = RCS_geom.plot_path_length_diff()
```

You can modify some parameters and the geometry is recomputed accordingly. In the following, the number of cells per arc is modified. The change of geometry can be observed on the cell length. If we change the pattern, we see the difference in trajectory for example. 

```python
print('Initial number of cell per arc', RCS_geom.nb_cell_arc)
print('Initial cell length', RCS_geom.cell_length)
RCS_geom.nb_cell_arc = 5
print('New number of cell per arc', RCS_geom.nb_cell_arc)
print('New cell length', RCS_geom.cell_length)

RCS_geom.pattern = ['BSC', 'BNC', 'BSC', 'BNC', 'BSC']
fig_traj2 = RCS_geom.plot_traj(t_traj)
```
