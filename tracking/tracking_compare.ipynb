Energy injection

```python
file_geo = '/mnt/c/muco/code/lattices/lattice_08_25/para_RCS1_gf_normal_qp__8_arc.json'
file_arc = '/mnt/c/muco/code/lattices/lattice_08_25/line_RCS1_gf_normal_qp_8_arc_v2.json'
    
line_arc = xt.Line.from_json(file_arc)
rcs = RingGeometry(file_geo)

p0_list = [63, 120, 180, 240, 300]
sigma_z_ref = 30.2e-12

method = 'var_ref'
RF_tuning = False
adapt_part = False
adapt_per = 80
plt_twiss = False
plt_sepa = True
```

```python
res = {}
res_adapt = {}
n_part=25000

for p0 in p0_list:
    print(f'----- Start study of {p0} GeV -----' )

    line_arc = xt.Line.from_json(file_arc)

    gamma0 = p0 * 1e9 / xp.MUON_MASS_EV
    line_arc.particle_ref = xp.Particles(
            p0c= p0 * 1e9, 
            q0=1, 
            mass0=xp.MUON_MASS_EV)

    fac_z = (rcs.injection_energy / p0 / 1e9)**0.25

    track = TrackingRCS(
        line = line_arc,
        rcs = rcs,
        n_turn_rcs = n_turn_rcs,
        n_part = n_part,
        nemitt_x = nemitt_x,
        nemitt_y = nemitt_y,
        sigma_z = sigma_z_ref * fac_z,
        t_acc = t_acc,
        method = method,
        RF_tuning = RF_tuning,
        adapt_part = adapt_part,
        adapt_percentile = adapt_per,
        plt_twiss = plt_twiss,
        plt_sepa = plt_sepa,
    )

    # track_adapt = TrackingRCS(
    #     line = line_arc,
    #     rcs = rcs,
    #     n_turn_rcs = n_turn_rcs,
    #     n_part = n_part,
    #     nemitt_x = nemitt_x,
    #     nemitt_y = nemitt_y,
    #     sigma_z = sigma_z_ref * fac_z,
    #     t_acc = t_acc,
    #     method = method,
    #     RF_tuning = RF_tuning,
    #     adapt_part = True,
    #     adapt_percentile = adapt_per,
    #     plt_twiss = plt_twiss,
    #     plt_sepa = plt_sepa,
    # )

    res[str(p0)] = track.track_res
    # res_adapt[str(p0)] = track_adapt.track_res

arc_length = line_arc.get_length()
mon_s = np.arange(0, n_turn_rcs * rcs.nb_arc) * arc_length
```

```python
def plot_Escan(res):
    fig, axes = plt.subplots(nrows=4, ncols=1, figsize=(6.4, 6.4))

    for key in res.keys():
        ex = res[key]['emitx']
        ey = res[key]['emity']
        es = res[key]['emits']
        tr = res[key]['transmission']

        axes[0].plot(mon_s * 1e-3, (ex - ex[0]) / ex[0] * 100)
        axes[1].plot(mon_s * 1e-3, (ey - ey[0]) / ey[0] * 100)
        axes[2].plot(mon_s * 1e-3, (es - es[0]) / es[0] * 100)
        axes[3].plot(mon_s * 1e-3, tr, label=f'{key} GeV')

    # Set log scale for the first three subplots

    axes[0].set_yscale('log')
    axes[0].yaxis.set_major_locator(LogLocator(base=10))
    axes[0].yaxis.set_major_formatter(ScalarFormatter())
    axes[0].yaxis.set_minor_formatter(NullFormatter())  # Use NullFormatter to hide minor tick labels

    # Set y-ticks for the first subplot
    axes[0].set_yticks([1, 10, 100, 500])

    # Set labels
    axes[0].set_ylabel(r'$\Delta \epsilon_x / \epsilon_x$ [%]')
    axes[1].set_ylabel(r'$\Delta \epsilon_y / \epsilon_y$ [%]')
    axes[2].set_ylabel(r'$\Delta \epsilon_s / \epsilon_s$ [%]')
    axes[3].set_ylabel('Transmission')
    axes[3].set_xlabel('s [km]')

    # Add a single legend for all subplots on the right
    fig.legend(
        loc='lower center', 
        bbox_to_anchor=(0.57, -0.028),
        ncol=3,
        )

    plt.tight_layout(rect=[0, 0.06, 1, 1])
    plt.show()

    return fig

fig = plot_Escan(res)
# fig_adapt = plot_Escan(res_adapt)

folder = '/mnt/c/muco/Results/figures/chap6/tracking_v2/RCS1_'
save_fig = True
ad = 'adapt' if adapt_part else 'no_adapt'
if save_fig:
    fig.savefig(folder + f'energy_scan_{ad}', bbox_inches='tight')
```

```python
file_geo = '/mnt/c/muco/code/lattices/lattice_08_25/para_RCS1_gf_normal_qp__8_arc.json'
file_arc = '/mnt/c/muco/code/lattices/lattice_08_25/line_RCS1_gf_normal_qp_8_arc_v2.json'
line_arc = xt.Line.from_json(file_arc)
rcs = RingGeometry(file_geo)

sz_list = np.array([30, 25, 20, 15, 10]) * 1e-12
res = {}
res_part = {}
n_part = 25000

for sz in sz_list:

    track = TrackingRCS(
        line = line_arc,
        rcs = rcs,
        n_turn_rcs = n_turn_rcs,
        n_part = n_part,
        nemitt_x = nemitt_x,
        nemitt_y = nemitt_y,
        sigma_z = sz,
        t_acc = t_acc,
        plt_sepa=True
    )

    res[str(sz*1e12)] = track.track_res
    res_part[str(sz*1e12)] = {}
    res_part[str(sz*1e12)]['part0'] = track.particles0
    res_part[str(sz*1e12)]['rf_bucket'] = track.rf_bucket

arc_length = line_arc.get_length()
mon_s = np.arange(1, n_turn_rcs*rcs.nb_arc+1) * arc_length
```

```python
fig_part0 = plt.figure(figsize=(6.4,4.8))
machine_rf_bucket = res_part[str(sz_list[0]*1e12)]['rf_bucket']
ufp = machine_rf_bucket.z_ufp
xx = np.linspace(machine_rf_bucket.z_left, machine_rf_bucket.z_right, 1000)
yy = machine_rf_bucket.separatrix(xx)
plt.plot((xx-machine_rf_bucket.z_sfp), yy, color='tab:orange')
plt.plot((xx-machine_rf_bucket.z_sfp), -yy, color='tab:orange')

for key in res_part.keys():
    es = res[key]['emits'][0]
    zeta = res_part[key]['part0'].zeta
    delta = res_part[key]['part0'].delta
    plt.scatter(zeta, delta, s=8, label=f'{es:.4f} eV.s')
plt.xlabel('z [m]')
plt.ylabel(r'$\delta$')
fig_part0.legend(
    loc='lower center', 
    bbox_to_anchor=(0.56, -0.06),
    ncol=3,
    markerscale=2
    )   
plt.tight_layout(rect=[0, 0.06, 1, 1])
plt.show()

folder = '/mnt/c/muco/Results/figures/chap6/tracking_v2/RCS1_'
fig_part0.savefig(folder+f'es_scan_no_adapt_part0',bbox_inches='tight')
```

```python
es0 = np.array([res[key]['emits'][0] for key in res.keys()])
dex = np.array([
    (res[key]['emitx'][-1] - res[key]['emitx'][0]) / res[key]['emitx'][0]
    for key in res.keys()
])

# Convert to percent
y = dex * 100

# Fit only positive values
mask = (es0 > 0) & (y > 0)

# log10(y) = m log10(x) + b
m, b = np.polyfit(np.log10(es0[mask]), np.log10(y[mask]), 1)
m_str, b_str = f'{m:.1f}', f'{b:.1f}'

# Smooth fitted curve
xfit = np.logspace(np.log10(es0[mask].min())*1.01, np.log10(es0[mask].max())*0.99, 200)
yfit = 10**b * xfit**m

fig_emit_energy_fit = plt.figure(figsize=(6.4, 4.8))
plt.scatter(es0, y, label='Simulation', marker='+', color='tab:orange', s=45)
plt.plot(
    xfit,
    yfit,
    lw=1.3,
    label=fr'Fit: $y = 10^{{{b_str}}} x^{{{m_str}}}$',
    color = 'tab:blue'
)
plt.xscale('log')
plt.yscale('log')
plt.xticks([0.002, 0.01, 0.02, 0.005],
           ['0.002', '0.01', '0.02', '0.005'])
plt.yticks([1, 5, 10, 50, 100, 300, 500],
           ['1', '5', '10', '50', '100', '300', '500'])
plt.ylim(1, 700)

plt.xlabel(r'$\epsilon_{s,0}$ [eVs]')
plt.ylabel(r'$\Delta \epsilon_x / \epsilon_x$ [%]')

plt.legend()
plt.tight_layout()
plt.show()

print(f"Power-law exponent = {m:.3f}")

if save_fig:
    fig_emit_energy_fit.savefig(folder + 'es_vs_ex_fit')
```
